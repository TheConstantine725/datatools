```python

with duckdb.connect() as duck:
    duck.sql(f"""install httpfs;
             load httpfs;
             attach if not exists '{DATABASES.joinpath("main.duckdb").as_posix()}' as duckdb_main;
             use duckdb_main;
             create schema if not exists taxi_trips;
             create table if not exists taxi_trips.high_volume_for_hire_trip_records as 
             select * from 'https://d37ci6vzurychx.cloudfront.net/trip-data/fhvhv_tripdata_2024-01.parquet' """)
```
