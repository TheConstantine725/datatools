```python
from pyspark.sql import SparkSession

session = (
    SparkSession.builder
    .master("spark://localhost:7077")
    .appName("Testing Pyspark")
    .config("spark.driver.host", "172.20.0.1")      # gateway = host from inside Docker
    .config("spark.driver.bindAddress", "0.0.0.0")  # listen on all interfaces
    .config("spark.driver.port", "4040")             # fixed port so workers can reach back
    .getOrCreate()
)

session.sql("select 1 as one").show()
```
